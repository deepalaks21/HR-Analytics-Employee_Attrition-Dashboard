# Department Analysis

```python
df['Department'].value_counts()
```

```python
plt.Figure(figsize=(10,6))
colors=['#24e0a1','#63d102','#24e04d']
sns.countplot(data=df, x='Department',hue='Department',palette=colors)
plt.title('Employees by Department')
plt.xlabel('Department')
plt.ylabel('Employee count')
plt.xticks(rotation=45)
plt.show()
```

# Attrition by Department

```python
pd.crosstab(df['Department'],df['Attrition'])
```

```python
colors=['#27d6f5','#0a70a3']
sns.countplot(data=df,x='Department',hue='Attrition',palette=colors)
plt.title('Attrition by Department')
plt.xticks(rotation=45)
plt.show()
```

```python
df['Jobrole'].value_counts()
```

```python
colors=['#9000b0','#da5ff5']
sns.countplot(data=df,y='Jobrole',hue='Attrition',palette=colors)
plt.title('Attrition by Jobrole')
plt.xlabel('Attrition counts')
plt.show()
```

```python
pd.crosstab(df['Gender'],df['Attrition'])
```

```python
colors=['#DB040B','#F7165A']
sns.countplot(data=df,x='Gender',hue='Attrition',palette=colors)
plt.title('Attrition by Gender')
plt.show()
```

```python
pd.crosstab(df['Overtime'],df['Attrition'])
```

```python
colors=["#067A86","#18F3F3"]
ax=sns.countplot(data=df,x='Overtime',hue='Attrition',palette=colors)
for container in ax.containers:
    ax.bar_label(container)
plt.title('Overtime vs Attrition')
plt.xlabel('Overtime')
plt.ylabel('Employee Count')
plt.show()
```
